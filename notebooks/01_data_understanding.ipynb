```python
from pathlib import Path
import pandas as pd

# Locate the project folder
PROJECT_DIR = Path.cwd().parent

# Dataset location
DATASET_PATH = PROJECT_DIR / "dataset" / "raw" / "traffic_accidents.csv"

# Check whether dataset exists
print("Dataset path:", DATASET_PATH)
print("Dataset exists:", DATASET_PATH.exists())

# Load first 200,000 records
if DATASET_PATH.exists():
    df = pd.read_csv(DATASET_PATH, nrows=200000, low_memory=False)

    print("\nDataset loaded successfully!")
    print("Dataset shape:", df.shape)

    display(df.head())
else:
    print("ERROR: Dataset file not found!")
```

```python
print("DATASET SHAPE:", df.shape)

print("\nALL COLUMN NAMES:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")

print("\nDATA TYPES:")
display(df.dtypes)
```

```python
# Standardize all column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)

print("Column names standardized successfully!\n")

for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

```python
# Calculate missing values
missing_count = df.isnull().sum()

# Calculate missing percentage
missing_percentage = (df.isnull().mean() * 100).round(2)

# Create missing values report
missing_report = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})

# Sort from highest to lowest missing percentage
missing_report = missing_report.sort_values(
    by="Missing_Percentage",
    ascending=False
)

print("MISSING VALUES ANALYSIS")
display(missing_report)

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Check duplicate rows
total_duplicates = df.duplicated().sum()

print("DUPLICATE RECORD ANALYSIS")
print("-------------------------")
print("Total duplicate rows:", total_duplicates)

# Check duplicate collision IDs
if "collision_id" in df.columns:
    duplicate_ids = df["collision_id"].duplicated().sum()
    print("Duplicate collision IDs:", duplicate_ids)
else:
    print("Collision ID column not found.")
```

```python
# NUMERICAL AND CATEGORICAL STATISTICAL ANALYSIS

print("=" * 60)
print("NUMERICAL COLUMN SUMMARY")
print("=" * 60)

# Identify numerical columns
numerical_columns = df.select_dtypes(include=["number"]).columns

display(df[numerical_columns].describe().T)

print("\n" + "=" * 60)
print("CATEGORICAL COLUMN SUMMARY")
print("=" * 60)

# Identify categorical columns
categorical_columns = df.select_dtypes(include=["object", "category"]).columns

display(df[categorical_columns].describe().T)
```

```python
# DATE AND TIME FEATURE ENGINEERING - CORRECTED VERSION

# Convert crash date
df["crash_date"] = pd.to_datetime(
    df["crash_date"],
    errors="coerce"
)

# Parse crash time
parsed_time = pd.to_datetime(
    df["crash_time"].astype(str),
    errors="coerce"
)

# Extract hour
df["crash_hour"] = parsed_time.dt.hour

# Display time without the unnecessary 1900-01-01 date
df["crash_time"] = parsed_time.dt.strftime("%H:%M:%S")

# Create date-related features
df["day_of_week"] = df["crash_date"].dt.day_name()
df["month"] = df["crash_date"].dt.month_name()
df["year"] = df["crash_date"].dt.year

# Identify weekends
df["is_weekend"] = df["crash_date"].dt.day_of_week >= 5

# Display analysis
print("=" * 60)
print("DATE AND TIME ANALYSIS")
print("=" * 60)

print("\nEarliest crash date:", df["crash_date"].min())
print("Latest crash date:", df["crash_date"].max())

print("\nValid crash dates:", df["crash_date"].notna().sum())
print("Invalid crash dates:", df["crash_date"].isna().sum())

print("\nValid crash times:", df["crash_hour"].notna().sum())
print("Invalid crash times:", df["crash_hour"].isna().sum())

print("\nEngineered Features:")

display(
    df[
        [
            "crash_date",
            "crash_time",
            "crash_hour",
            "day_of_week",
            "month",
            "year",
            "is_weekend"
        ]
    ].head(10)
)
```
